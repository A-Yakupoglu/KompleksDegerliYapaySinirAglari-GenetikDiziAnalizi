## Veri Setinin Oluşturulması

- Bu çalışmada, insan (Homo sapiens) protein kinaz ve G-protein bağlı reseptör (GPCR) ailelerinden oluşan bir veri kümesi kullanılmıştır. İnsan genomunda çok sayıda protein kinaz ve GPCR bulunması, bu iki ailenin derin öğrenme tabanlı sınıflandırma için yeterli çeşitlilik sağlamasına olanak vermektedir. Veri kümesi, model girdilerindeki uzunluk farkını sınırlamak amacıyla 300–1500 amino asit uzunluğundaki proteinlerle sınırlandırılmıştır.

#### 1. Erişim Numaralarının Belirlenmesi

- Protein dizileri, insan taksonomisine ait ve küratörlü (reviewed) UniProt kayıtlarından alınmıştır. Protein kinazların belirlenmesinde EC 2.7.10–2.7.13 sınıfları, GPCR'lerin belirlenmesinde ise küratörlü KW-0297 anahtar kelimesi kullanılmıştır. Sorgular 2024 yılı itibarıyla gerçekleştirilmiş; veri kümesi, güncel UniProt kayıtlarını yansıtmak amacıyla UniProt 2026_02 sürümü kullanılarak güncellenmiştir.

- Her kayıt için protein dizisi ve RefSeq protein erişim numaraları alınmıştır. Birden fazla RefSeq kaydı bulunan proteinlerde NP_ kayıtları ve izoformu gösteren adaylar öncelikli olarak değerlendirilmiş, ancak kesin seçim sonraki aşamada protein dizisinin CDS-DNA çevirisiyle karşılaştırılmasıyla yapılmıştır.

- Bu aşamada yalnızca protein dizisi ve en az bir RefSeq protein erişim numarası bulunan kayıtlar havuza alınmıştır. Böylece 767 GPCR ve 432 protein kinazdan oluşan bir aday havuzu elde edilmiştir. Domain, fragman ve amino asit alfabesi gibi göstergeler, sonraki değerlendirmelerde kullanılmak üzere kaydedilmiştir. İki protein ailesine ait erişim numaraları arasında çakışma bulunmadığı görülmüştür.

#### 2. CDS Dizilerinin Elde Edilmesi ve Doğrulanması

- Aday proteinlerin kodlayıcı DNA (CDS) dizilerini elde etmek için RefSeq protein kayıtları ile ilgili transkript kayıtları NCBI üzerinden eşleştirilmiştir. Bu amaçla GenPept kayıtlarındaki coded bilgileri kullanılarak ilgili transkriptler ve kodlayıcı bölgeler belirlenmiştir. CDS dizileri NCBI'dan toplu olarak alınmış; gerekli durumlarda kayıt üzerindeki koordinatlar kullanılarak elde edilmiştir.

- Bir proteine birden fazla RefSeq adayı karşılık gelebildiğinden, aday CDS dizileri sırayla çevrilmiş ve elde edilen amino asit dizisi UniProt dizisiyle karşılaştırılmıştır. UniProt dizisini birebir karşılayan aday dizi doğrulanmış ve seçilmiştir. Böylece protein ve DNA dizilerinin aynı biyolojik ürünü temsil edip etmediği kontrol edilmiştir.

- Bu işlem sonucunda kaydın 1,154'ü doğrulanmıştır. Hiçbir kayıtta CDS'nin tamamen elde edilememesi söz konusu olmamıştır. Doğrulama aşamasında kayıtlar elenmemiş, sonuçlar sonraki filtreleme aşamasında kullanılmak üzere kaydedilmiştir.

#### 3. Filtreleme ve Veri Kümesinin Oluşturulması

- Filtreleme işlemi veri indirme ve doğrulama aşamalarından ayrı olarak gerçekleştirilmiştir. Bu yaklaşım, farklı filtrelerin veri indirme işlemi tekrarlanmadan test edilmesine olanak sağlamıştır. Filtreleme sonrasında 727 GPCR ve 427 protein kinaz kaydı örnekleme için uygun bulunmuştur.

## 1. Adım: Erişim numarası havuzunun oluşturulması ve denetimi


```python
# =============================================================================
# ADIM 1 - GPCR / PROTEIN KINAZ ERISIM NUMARASI HAVUZU
# -----------------------------------------------------------------------------
# 
#   GPCR  -> Swiss-Prot kuratorlu anahtar kelime KW-0297
#   KINAZ -> EC 2.7.10-2.7.13 (protein kinaz alt siniflari)
#   Ortak -> insan (9606), reviewed:true, 300-1500 aa
#
# NOT: Bu erişim numaraları tez veri seti için kullanılmıştır. UniProt değişken olduğundan, aynı sorgu
# daha sonraki ay ve yıllarda farklı olabileceğinden , o zamanki verilerin  kaydı bu dosyadır.
# =============================================================================

import io
import os
import re
import sys
import time
from datetime import datetime

import pandas as pd
import requests

# ----------------------------- KONFIGURASYON ---------------------------------
TAXONOMY = "9606"
LENGTH_MIN, LENGTH_MAX = 300, 1500
PK_EC = ["2.7.10.*", "2.7.11.*", "2.7.12.*", "2.7.13.*"]
GPCR_KEYWORD = "KW-0297"

# Domain imzalari
GPCR_SINIF = [("A",        {"PF00001", "IPR000276", "IPR017452"}),
              ("B",        {"PF00002", "IPR000832"}),
              ("C",        {"PF00003", "IPR000337"}),
              ("Frizzled", {"PF01534"}),
              ("TAS2R",    {"PF05296"})]
PK_DOMAIN = {"PF00069", "PF07714",                    # Pfam
             "IPR000719", "IPR001245", "IPR011009"}   # InterPro
GPCR_DOMAIN = {t for _, k in GPCR_SINIF for t in k}
# Kinaz alt sinifi EC'den okunur 
PK_EC_AD = {"2.7.10": "Tirozin", "2.7.11": "Ser-Thr",
            "2.7.12": "Cift-ozgul", "2.7.13": "Histidin"}

FIELDS = ("accession,id,protein_name,gene_names,length,ec,"
          "xref_pfam,xref_interpro,xref_refseq,sequence,fragment")

OUT_XLSX = "1_havuz.xlsx"
LOG = "boru_hatti.log"

STANDART_AA = set("ACDEFGHIKLMNPQRSTVWY")
RE_PROT_ACC = re.compile(r"^(NP|XP)_\d+(\.\d+)?$")

HAVUZ_KOLON = ["Uniprot_Accession", "Entry_ID", "Gene_Name", "Protein_Name",
               "label", "alt_sinif", "Length", "EC", "Pfam", "InterPro",
               "n_aday", "NP_adaylari", "sorun", "AA_Sequence"]


# ----------------------------- GUNLUK ----------------------------------------
class Gunluk:


    def __init__(self, adim: str):
        self.adim, self.satirlar = adim, []

    def __call__(self, metin: str = ""):
        print(metin)
        self.satirlar.append(str(metin))

    def yaz(self, yol: str = LOG):
        with open(yol, "a", encoding="utf-8") as f:
            f.write("\n" + "=" * 64 + "\n")
            f.write(f"{self.adim} | {datetime.now():%Y-%m-%d %H:%M:%S}\n")
            f.write("\n".join(self.satirlar) + "\n")
        print(f"Gunluk : {yol} (eklendi)")


# ----------------------------- SORGULAR --------------------------------------
def kinaz_sorgusu() -> str:
    ec = " OR ".join(f"ec:{e}" for e in PK_EC)
    return (f"({ec}) AND taxonomy_id:{TAXONOMY} AND reviewed:true "
            f"AND length:[{LENGTH_MIN} TO {LENGTH_MAX}]")


def gpcr_sorgusu() -> str:
    return (f"keyword:{GPCR_KEYWORD} AND taxonomy_id:{TAXONOMY} AND reviewed:true "
            f"AND length:[{LENGTH_MIN} TO {LENGTH_MAX}]")


# ----------------------------- INDIRME ---------------------------------------
def uniprot_cek(sorgu: str, etiket: str):

    url = "https://rest.uniprot.org/uniprotkb/search"
    params = {"query": sorgu, "format": "tsv", "fields": FIELDS, "size": 500}
    parcalar, sonraki, bekleme, surum = [], url, 0, {}
    print(f"{etiket} indiriliyor", end="", flush=True)

    while sonraki:
        r = requests.get(sonraki, params=params if sonraki == url else None, timeout=60)
        if r.status_code == 429:
            bekleme += 1
            if bekleme > 5:
                print("\n  HATA: UniProt istek siniri asildi.")
                break
            time.sleep(2.0 * bekleme)
            continue
        if r.status_code != 200:
            print(f"\n  HATA {r.status_code}: {r.text[:200]}")
            break
        bekleme = 0
        if not surum:
            surum = {k: v for k, v in r.headers.items()
                     if k.lower().startswith("x-uniprot")}
        df = pd.read_csv(io.StringIO(r.text), sep="\t")
        if not df.empty:
            parcalar.append(df)
        sonraki = r.links["next"]["url"] if "next" in r.links else None
        if sonraki:
            print(".", end="", flush=True)
            time.sleep(0.2)

    if not parcalar:
        print("  BOS!")
        return pd.DataFrame(), "bilinmiyor"
    out = pd.concat(parcalar, ignore_index=True)
    print(f" tamam ({len(out)} kayit).")
    metin = "; ".join(f"{k}={v}" for k, v in surum.items()) if surum else "bilinmiyor"
    return out, metin


# ----------------------------- KOLON NORMALIZASYONU --------------------------
HARITA = {  # standart ad: (aday basliklar, bulanik eslesme anahtar)
    "Uniprot_Accession": (["Entry", "accession"], "entry"),
    "Entry_ID":          (["Entry Name", "id"], "entry name"),
    "Protein_Name":      (["Protein names", "protein_name"], "protein name"),
    "Gene_Name":         (["Gene Names", "gene_names"], "gene"),
    "Length":            (["Length", "length"], "length"),
    "EC":                (["EC number", "ec"], "ec"),
    "Pfam":              (["Pfam", "xref_pfam"], "pfam"),
    "InterPro":          (["InterPro", "xref_interpro"], "interpro"),
    "RefSeq":            (["RefSeq", "xref_refseq"], "refseq"),
    "AA_Sequence":       (["Sequence", "sequence"], "sequence"),
    "Fragment":          (["Fragment", "fragment"], "fragment"),
}
KRITIK = {"Uniprot_Accession", "AA_Sequence", "RefSeq"}


def kolonlari_normalize_et(df: pd.DataFrame, etiket: str) -> pd.DataFrame:

    out, eksik, bulanik, bos = pd.DataFrame(index=df.index), [], [], []
    for std, (adaylar, anahtar) in HARITA.items():
        kaynak = next((c for c in adaylar if c in df.columns), "")
        if not kaynak:
            kaynak = next((c for c in df.columns if anahtar in c.lower()), "")
            if kaynak:
                bulanik.append(f"{std}<-{kaynak}")
        if not kaynak:
            (eksik if std in KRITIK else bos).append(std)
        out[std] = df[kaynak].astype(str) if kaynak else ""
    if eksik:
        print(f"\n!! DURDURULDU [{etiket}]: kritik kolon(lar) gelmedi -> {eksik}")
        print(f"   Gelen kolonlar: {list(df.columns)}")
        print("   UniProt alan adlari degismis olabilir; FIELDS sabitini kontrol edin.")
        sys.exit(1)
    if bulanik:
        print(f"   ! [{etiket}] bulanik kolon eslesmesi: {bulanik}")
    if bos:
        print(f"   ! [{etiket}] gelmeyen (kritik olmayan) kolon: {bos}")

    out["Length"] = pd.to_numeric(out["Length"], errors="coerce")
    out["AA_Sequence"] = (out["AA_Sequence"].str.upper()
                          .str.replace(r"[^A-Z]", "", regex=True)
                          .replace({"NAN": ""}))
    return out


# ----------------------------- REFSEQ ADAYLARI -------------------------------
def refseq_adaylari(hucre) -> list:

    if pd.isna(hucre) or str(hucre).strip().lower() in ("", "nan"):
        return []
    adaylar = []
    for parca in str(hucre).split(";"):
        parca = parca.strip()
        if not parca:
            continue
        m_iso = re.search(r"\[([^\]]+)\]", parca)
        etiket = m_iso.group(1).strip() if m_iso else ""
        acc = re.sub(r"\s*\[.*?\]\s*", "", parca).strip()
        if RE_PROT_ACC.match(acc):
            adaylar.append((0 if acc.startswith("NP_") else 1,
                            0 if etiket.endswith("-1") else (1 if not etiket else 2),
                            acc))
    adaylar.sort(key=lambda x: (x[0], x[1]))
    return [a[2] for a in adaylar]


# ----------------------------- KALITE - ALT SINIF ----------------------------
def _jetonlar(h) -> set:
    return set() if pd.isna(h) else {t.strip() for t in re.split(r"[;,\s]+", str(h)) if t.strip()}


def gpcr_alt_sinif(jetonlar: set) -> str:
    for ad, kume in GPCR_SINIF:
        if jetonlar & kume:
            return ad
    return "-"


def kinaz_alt_sinif(ec_hucre) -> str:
    for e in _jetonlar(ec_hucre):
        ad = PK_EC_AD.get(".".join(e.split(".")[:3]))
        if ad:
            return ad
    return "-"


def hazirla(df: pd.DataFrame, sinif: str) -> pd.DataFrame:

    out = df.copy()
    imza = PK_DOMAIN if sinif == "KINASE" else GPCR_DOMAIN
    jeton = (out["Pfam"] + ";" + out["InterPro"]).apply(_jetonlar)

    domain_var = jeton.apply(lambda j: bool(j & imza))
    fragman = out["Fragment"].str.strip().str.lower().isin(["fragment", "fragments"])
    alfabe_ok = out["AA_Sequence"].apply(lambda s: bool(s) and set(s) <= STANDART_AA)
    
    
    uzunluk_ok = out["AA_Sequence"].str.len().eq(out["Length"])

    out["sorun"] = [";".join(
        (["domain_imzasi_yok"] if not d else []) +
        (["fragman"] if f else []) +
        (["standart_disi_harf"] if not a else []) +
        (["uzunluk_uyusmuyor"] if not u else []))
        for d, f, a, u in zip(domain_var, fragman, alfabe_ok, uzunluk_ok)]

    out["alt_sinif"] = (jeton.apply(gpcr_alt_sinif) if sinif == "GPCR"
                        else out["EC"].apply(kinaz_alt_sinif))
    out["NP_adaylari"] = out["RefSeq"].apply(refseq_adaylari)
    out["n_aday"] = out["NP_adaylari"].apply(len)
    out["label"] = sinif

    out["_dizi_var"] = out["AA_Sequence"].str.len() > 0
    out["_np_var"] = out["n_aday"] > 0
    out["HAVUZDA"] = out["_dizi_var"] & out["_np_var"]
    out["_domain_var"] = domain_var
    return out


def huni(df: pd.DataFrame, sinif: str) -> pd.DataFrame:
    ic = df[df.HAVUZDA]
    return pd.DataFrame({
        "Sinif": sinif,
        "Asama": ["UniProt ham kayit",
                  "  dizi yok (dusen)",
                  "  NP_ adayi yok (dusen)",
                  "HAVUZ",
                  "  domain imzasi tasiyan",
                  "  sorun isaretli"],
        "Sayi": [len(df),
                 int((~df["_dizi_var"]).sum()),
                 int((df["_dizi_var"] & ~df["_np_var"]).sum()),
                 len(ic),
                 int(ic["_domain_var"].sum()),
                 int(ic["sorun"].ne("").sum())],
    })


# ----------------------------- ANA AKIŞ --------------------------------------
def main():
    g = Gunluk("ADIM 1 - HAVUZ")
    gpcr_ham, surum_g = uniprot_cek(gpcr_sorgusu(), "GPCR")
    kin_ham, surum_k = uniprot_cek(kinaz_sorgusu(), "Protein Kinaz")
    if gpcr_ham.empty or kin_ham.empty:
        print("Veri cekilemedi; listelerden biri bos. Sorgu/baglanti kontrol edin.")
        return

    g(f"UniProt surumu : {surum_g if surum_g != 'bilinmiyor' else surum_k}")
    g(f"GPCR sorgusu   : {gpcr_sorgusu()}")
    g(f"Kinaz sorgusu  : {kinaz_sorgusu()}")
    g(f"Alanlar        : {FIELDS}")
    g("")

    gpcr = hazirla(kolonlari_normalize_et(gpcr_ham, "GPCR"), "GPCR")
    kin = hazirla(kolonlari_normalize_et(kin_ham, "KINAZ"), "KINASE")

    # Iki listede birden gorunen kayit ikisinden de Çıkar
    cakisma = sorted(set(gpcr["Uniprot_Accession"]) & set(kin["Uniprot_Accession"]))
    if cakisma:
        gpcr = gpcr[~gpcr["Uniprot_Accession"].isin(cakisma)]
        kin = kin[~kin["Uniprot_Accession"].isin(cakisma)]

    tablo = pd.concat([huni(gpcr, "GPCR"), huni(kin, "KINASE")], ignore_index=True)
    tum = pd.concat([gpcr, kin], ignore_index=True) \
            .drop_duplicates(subset=["Uniprot_Accession"], keep="first")
    havuz = tum[tum["HAVUZDA"]].copy()
    havuz["NP_adaylari"] = havuz["NP_adaylari"].apply("|".join)

    havuz[HAVUZ_KOLON].to_excel(OUT_XLSX, sheet_name="HAVUZ", index=False)

    g(tablo.to_string(index=False))
    g("")
    g("Alt sinif dagilimi (havuz):")
    g(havuz.groupby(["label", "alt_sinif"]).size().rename("n").to_string())
    g("")
    g("Uzunluk (aa) dagilimi:")
    g(havuz.groupby("label")["Length"]
      .describe(percentiles=[.25, .5, .75])[["count", "min", "25%", "50%", "75%", "max"]]
      .round(1).to_string())
    g("")
    g(f"Sinif cakismasi (ikisinden de cikarildi): {len(cakisma)}"
      + (" -> " + ", ".join(cakisma[:20]) if cakisma else ""))
    g(f"Cikti  : {os.path.abspath(OUT_XLSX)}")
    g("Not: 'sorun' sutunu eleme yapmaz; eleme karari ADIM 3'tedir.")
    g.yaz()


if __name__ == "__main__":
    main()
```

## 2. Adım: CDS dizilerinin elde edilmesi ve doğrulanması


```python
# =============================================================================
# ADIM 2 - CDS (DNA) INDIRME ve PROTEIN - DNA DOGRULAMA
# -----------------------------------------------------------------------------

# =============================================================================

import os
import re
import time
from datetime import datetime

import pandas as pd
import requests
from Bio.Seq import Seq

IN_XLSX  = "1_havuz.xlsx"
OUT_XLSX = "2_cds.xlsx"
LOG      = "boru_hatti.log"

EFETCH    = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"
API_KEY   = os.environ.get("NCBI_API_KEY", "")
EPOSTA    = os.environ.get("NCBI_EMAIL", "abdullah.yakupoglu17@atauni.edu.tr")
SLEEP     = 0.11 if API_KEY else 0.34   # anahtarsiz 3 istek/sn 
TIMEOUT   = (10, 90)
MAX_RETRY = 4
BATCH     = 100

KISMI_ESIK = 0.98   

VERI_KOLON = ["Uniprot_Accession", "Entry_ID", "Gene_Name", "Protein_Name",
              "label", "alt_sinif", "Length", "sorun",
              "n_aday", "denenen_aday", "secilen_NP", "secilen_NM", "kaynak",
              "cds_durum", "benzerlik", "cds_len", "CDS_DOGRU",
              "AA_Sequence", "cds_dna"]


# ----------------------------- GUNLUK ----------------------------------------
class Gunluk:
    def __init__(self, adim: str):
        self.adim, self.satirlar = adim, []

    def __call__(self, metin: str = ""):
        print(metin)
        self.satirlar.append(str(metin))

    def yaz(self, yol: str = LOG):
        with open(yol, "a", encoding="utf-8") as f:
            f.write("\n" + "=" * 64 + "\n")
            f.write(f"{self.adim} | {datetime.now():%Y-%m-%d %H:%M:%S}\n")
            f.write("\n".join(self.satirlar) + "\n")
        print(f"Gunluk : {yol} (eklendi)")


# ----------------------------- HTTP ------------------------------------------
def _istek(params: dict) -> str:
    p = dict(params, tool="gpcr-kinaz-pipeline", email=EPOSTA)
    if API_KEY:
        p["api_key"] = API_KEY
    son = None
    for deneme in range(1, MAX_RETRY + 1):
        try:
            r = requests.post(EFETCH, data=p, timeout=TIMEOUT)
            if r.status_code in (429, 500, 502, 503, 504):
                son = f"HTTP {r.status_code}"
                time.sleep(SLEEP * (2 ** deneme))
                continue
            r.raise_for_status()
            return r.text
        except Exception as e:
            son = f"{type(e).__name__}: {e}"
            time.sleep(SLEEP * (2 ** deneme))
    print(f"\n    ! istek basarisiz: {son}")
    return ""


# -----------------------------  / KARSILAŞTIRMA ------------------------
def kok(acc: str) -> str:
    """Surum numarasini atar: 'NP_005219.2' -> 'NP_005219'."""
    return str(acc).split(".")[0].strip()


def cevir(dna: str) -> str:
    d = re.sub(r"[^ACGTN]", "", str(dna).upper())
    d = d[: (len(d) // 3) * 3]
    if len(d) < 3:
        return ""
    try:
        return str(Seq(d).translate(to_stop=False))
    except Exception:
        return ""


def esles(aa_ref: str, cds: str):

    if not cds:
        return "CDS_YOK", 0.0
    aa = cevir(cds)
    if len(aa) == len(aa_ref) + 1 and aa.endswith("*"):
        aa = aa[:-1]
    if not aa:
        return "CDS_YOK", 0.0
    if aa == aa_ref:
        return "TAM_ESLESME", 1.0
    if len(aa) == len(aa_ref):
        uyan = sum(1 for a, b in zip(aa, aa_ref) if a == b or (a == "*" and b == "U"))
        oran = uyan / len(aa_ref)
        if oran == 1.0:
            return "TAM_ESLESME", 1.0            
        return ("KISMI" if oran >= KISMI_ESIK else "UYUSMAZLIK"), oran
    ortak = sum(1 for a, b in zip(aa, aa_ref) if a == b)
    return "IZOFORM_FARKI", ortak / max(len(aa_ref), 1)


def cds_temiz_mi(cds: str) -> bool:
    d = str(cds).upper()
    return bool(d) and len(d) % 3 == 0 and set(d) <= set("ACGT")


# ----------------------------- A) GENPEPT  /coded_by -----------------------
def coded_by_coz(kod: str):
    """/coded_by degerini (nuccore_acc, [(bas,son)...], iplik) olarak ayristirir."""
    kod, iplik = kod.strip(), 1
    if kod.startswith("complement(") and kod.endswith(")"):
        iplik, kod = 2, kod[len("complement("):-1].strip()
    govde = kod[len("join("):-1] if (kod.startswith("join(") and kod.endswith(")")) else kod
    acc, araliklar = "", []
    for parca in govde.split(","):
        m = re.match(r"([A-Z]{1,4}_?\d+(?:\.\d+)?):<?(\d+)\.\.>?(\d+)", parca.strip())
        if m:
            acc = acc or m.group(1)
            a, b = int(m.group(2)), int(m.group(3))
            araliklar.append((min(a, b), max(a, b)))
    if iplik == 2:
        araliklar = araliklar[::-1]
    return (acc, araliklar, iplik) if acc and araliklar else None


def genpept_ayristir(metin: str) -> dict:

    sonuc = {}
    for blok in metin.split("\n//"):
        mv = re.search(r"^VERSION\s+(\S+)", blok, re.M)
        mc = re.search(r'/coded_by="([^"]+)"', blok, re.S)
        if not (mv and mc):
            continue
        
        coz = coded_by_coz(re.sub(r"\s+", "", mc.group(1)))
        if coz:
            sonuc[kok(mv.group(1))] = coz
    return sonuc


def toplu_genpept(np_listesi, g) -> dict:
    havuz, tekil = {}, sorted({kok(x) for x in np_listesi if x})
    print(f"[A] GenPept: {len(tekil)} protein | ~{-(-len(tekil) // BATCH)} istek")
    for i in range(0, len(tekil), BATCH):
        havuz.update(genpept_ayristir(
            _istek({"db": "protein", "id": ",".join(tekil[i:i + BATCH]),
                    "rettype": "gp", "retmode": "text"})))
        print(f"    {min(i + BATCH, len(tekil))}/{len(tekil)}", end="\r", flush=True)
        time.sleep(SLEEP)
    print()
    g(f"[A] GenPept /coded_by cozulen: {len(havuz)}/{len(tekil)} protein")
    return havuz


# ----------------------------- B) NM_  CDS (toplu) -------------------------
def fasta_cds_na_ayristir(metin: str) -> dict:

    sonuc, pid, parcalar = {}, None, []
    for satir in metin.splitlines():
        if satir.startswith(">"):
            if pid and parcalar:
                sonuc[pid] = "".join(parcalar)
            parcalar = []
            m = re.search(r"\[protein_id=([A-Z]{2}_\d+(?:\.\d+)?)\]", satir)
            pid = kok(m.group(1)) if m else None
        elif pid:
            parcalar.append(satir.strip())
    if pid and parcalar:
        sonuc[pid] = "".join(parcalar)
    return sonuc


def toplu_cds(nm_listesi, g) -> dict:
    havuz, tekil = {}, sorted({x for x in nm_listesi if x})
    print(f"[B] CDS: {len(tekil)} transkript | ~{-(-len(tekil) // BATCH)} istek")
    for i in range(0, len(tekil), BATCH):
        havuz.update(fasta_cds_na_ayristir(
            _istek({"db": "nuccore", "id": ",".join(tekil[i:i + BATCH]),
                    "rettype": "fasta_cds_na", "retmode": "text"})))
        print(f"    {min(i + BATCH, len(tekil))}/{len(tekil)}", end="\r", flush=True)
        time.sleep(SLEEP)
    print()
    g(f"[B] Toplu CDS gelen protein   : {len(havuz)}/{len(tekil)} transkript")
    return havuz


# ----------------------------- C) YEDEK: koordinatla kesme -------------------
def koordinatla_kes(nuccore: str, araliklar, iplik: int) -> str:
    parcalar = []
    for bas, son in araliklar:
        fa = _istek({"db": "nuccore", "id": nuccore, "rettype": "fasta",
                     "retmode": "text", "seq_start": str(bas), "seq_stop": str(son),
                     "strand": str(iplik)})
        parcalar.append("".join(fa.splitlines()[1:]) if fa.startswith(">") else "")
        time.sleep(SLEEP)
    return "".join(parcalar)


# ----------------------------- ANA AKIŞ --------------------------------------
def main():
    g = Gunluk("ADIM 2 - CDS")
    havuz = pd.read_excel(IN_XLSX, sheet_name="HAVUZ").reset_index(drop=True)
    havuz["adaylar"] = havuz["NP_adaylari"].apply(
        lambda s: [p.strip() for p in str(s).split("|") if p.strip()])
    toplam_aday = int(sum(len(a) for a in havuz["adaylar"]))
    g(f"Girdi          : {IN_XLSX} | {len(havuz)} kayit, {toplam_aday} NP_ adayi")
    g(f"NCBI API anahtari: {'var' if API_KEY else 'yok (3 istek/sn)'}")

    gp = toplu_genpept([np for L in havuz["adaylar"] for np in L], g)   # A
    cds_havuzu = toplu_cds([v[0] for v in gp.values()], g)              # B

    kayitlar, yedek = [], 0
    for i, sat in havuz.iterrows():
        aa_ref = str(sat["AA_Sequence"])
        en_iyi = {"secilen_NP": "", "secilen_NM": "", "cds_dna": "", "kaynak": "",
                  "cds_durum": "ADAY_YOK", "benzerlik": 0.0, "denenen_aday": 0}
        for np_acc in sat["adaylar"]:
            k = kok(np_acc)
            bilgi = gp.get(k)
            cds, kaynak = cds_havuzu.get(k, ""), "toplu"
            if not cds and bilgi:                                       # C
                cds, kaynak = koordinatla_kes(*bilgi), "koordinat"
                yedek += 1
            durum, benz = esles(aa_ref, cds)
            en_iyi["denenen_aday"] += 1
            if benz > en_iyi["benzerlik"] or en_iyi["cds_durum"] == "ADAY_YOK":
                en_iyi.update(secilen_NP=np_acc, secilen_NM=bilgi[0] if bilgi else "",
                              cds_dna=cds, kaynak=kaynak,
                              cds_durum=durum, benzerlik=round(benz, 4))
            if durum == "TAM_ESLESME":
                break
        kayitlar.append(en_iyi)
        if (i + 1) % 25 == 0:
            print(f"  cozum: {i + 1}/{len(havuz)}", end="\r", flush=True)
    print()

    veri = pd.concat([havuz.drop(columns=["adaylar", "NP_adaylari"]),
                      pd.DataFrame(kayitlar)], axis=1)
    veri["cds_len"] = veri["cds_dna"].str.len()
    veri["CDS_DOGRU"] = veri["cds_durum"].eq("TAM_ESLESME") & veri["cds_dna"].apply(cds_temiz_mi)

    veri[VERI_KOLON].to_excel(OUT_XLSX, sheet_name="VERI", index=False)

    dogru = veri[veri.CDS_DOGRU]
    g(f"[C] Koordinatla kesmeye dusen aday: {yedek}")
    g("")
    g("Durum dagilimi:")
    g(pd.crosstab(veri["label"], veri["cds_durum"]).to_string())
    g("")
    g("Dogrulama orani:")
    g(veri.groupby("label")["CDS_DOGRU"]
      .agg(dogrulanan="sum", toplam="count")
      .assign(oran=lambda d: (d.dogrulanan / d.toplam).round(3)).to_string())
    g("")
    
    g("Cok adayli NP_ stratejisinin katkisi (dogrulanan kayitlar icinde):")
    g(f"  ilk adayla cozulen        : {int((dogru['denenen_aday'] == 1).sum())}")
    g(f"  ikinci veya sonraki adayla: {int((dogru['denenen_aday'] > 1).sum())}")
    g(f"  koordinatla kesilerek     : {int((dogru['kaynak'] == 'koordinat').sum())}")
    g("")
    g(f"Cikti  : {os.path.abspath(OUT_XLSX)}")
    g.yaz()


if __name__ == "__main__":
    main()
```

## 3. Adım: Örneklem seçimi ve veri kümesinin oluşturulması


```python
# =============================================================================
# ADIM 3 - FILTRE KARARI + ORNEKLEME + EGITIM/TEST BOLME
# -----------------------------------------------------------------------------
# TUM ELEME KARARLARI BURADADIR. 
#
# =============================================================================

import os
from datetime import datetime

import numpy as np
import pandas as pd

IN_XLSX  = "2_cds.xlsx"
OUT_XLSX = "3_veri_seti.xlsx"
LOG      = "boru_hatti.log"

N_EGITIM = 200          
N_TEST   = 100          
SEED     = 42

# ---- FİLTRE ANAHTARLARI  ----
SART_CDS_DOGRU   = True     
SART_DOMAIN      = False    
SART_ALFABE      = False   
SART_FRAGMAN_HRC = False   

# ---- ÖRNEKLEME ----
TABAKALI_UZUNLUK = False   
TABAKA_SAYISI    = 4

# ---- REDUNDANS ----
REDUNDANS_OLC    = True    
REDUNDANS_UYGULA = False   
REDUNDANS_ESIK   = 0.90
K = 5

FASTA_YAZ = True           

VERI_KOLON = ["Uniprot_Accession", "Entry_ID", "Gene_Name", "Protein_Name",
              "label", "alt_sinif", "split", "Length", "secilen_NP",
              "secilen_NM", "cds_len", "AA_Sequence", "cds_dna"]

SORUN_ETIKET = {"SART_DOMAIN": "domain_imzasi_yok",
                "SART_ALFABE": "standart_disi_harf",
                "SART_FRAGMAN_HRC": "fragman"}


# ----------------------------- GUNLUK ----------------------------------------
class Gunluk:
    def __init__(self, adim: str):
        self.adim, self.satirlar = adim, []

    def __call__(self, metin: str = ""):
        print(metin)
        self.satirlar.append(str(metin))

    def yaz(self, yol: str = LOG):
        with open(yol, "a", encoding="utf-8") as f:
            f.write("\n" + "=" * 64 + "\n")
            f.write(f"{self.adim} | {datetime.now():%Y-%m-%d %H:%M:%S}\n")
            f.write("\n".join(self.satirlar) + "\n")
        print(f"Gunluk : {yol} (eklendi)")


# ----------------------------- REDUNDANS -------------------------------------
def kmer(s: str, k: int = K) -> frozenset:
    return frozenset(s[i:i + k] for i in range(len(s) - k + 1))


def jaccard(a: frozenset, b: frozenset) -> float:
    if not a or not b:
        return 0.0
    ks = len(a & b)
    return ks / (len(a) + len(b) - ks)


def redundans_olc(diziler: list) -> np.ndarray:
    """Her dizi icin diğer tum dizilerle olan EN YUKSEK Jaccard degeri."""
    km = [kmer(s) for s in diziler]
    en = np.zeros(len(km))
    for i in range(len(km)):
        for j in range(i + 1, len(km)):
            v = jaccard(km[i], km[j])
            en[i] = max(en[i], v)
            en[j] = max(en[j], v)
    return en


def redundans_azalt(df: pd.DataFrame, esik: float, g) -> pd.DataFrame:
   
    d = df.assign(_u=df["AA_Sequence"].str.len()).sort_values("_u", ascending=False)
    km = {i: kmer(s) for i, s in d["AA_Sequence"].items()}
    tems = []
    for i in d.index:
        if not any(jaccard(km[i], km[t]) >= esik for t in tems):
            tems.append(i)
    g(f"    redundans indirgeme: {len(d)} -> {len(tems)} (elenen {len(d) - len(tems)})")
    return d.loc[tems].drop(columns="_u").reset_index(drop=True)


# ----------------------------- FILTRE ----------------------------------------
def filtre_uygula(df: pd.DataFrame, g) -> pd.DataFrame:
    sorun = df["sorun"].fillna("").astype(str)
    m = pd.Series(True, index=df.index)
    g(f"    {'baslangic':<22} -> {int(m.sum())}")
    if SART_CDS_DOGRU:
        m &= df["CDS_DOGRU"].astype(bool)
        g(f"    {'CDS dogrulanmis':<22} -> {int(m.sum())}")
    for anahtar, etiket in SORUN_ETIKET.items():
        if globals()[anahtar]:
            m &= ~sorun.str.contains(etiket)
            g(f"    {etiket + ' haric':<22} -> {int(m.sum())}")
    return df[m].copy()


# ----------------------------- BÖLME -----------------------------------------
def _kota(paylar: np.ndarray, toplam: int) -> np.ndarray:

    ham = paylar * toplam
    kota = np.floor(ham).astype(int)
    for i in np.argsort(-(ham - kota))[:toplam - kota.sum()]:
        kota[i] += 1
    return kota


def bol(df: pd.DataFrame, sinif: str, seed: int, g):
    gerekli = N_EGITIM + N_TEST
    if len(df) < gerekli:
        raise ValueError(
            f"[{sinif}] Yetersiz kayit: {len(df)} < {gerekli}.\n"
            f"  Cozum onerileri: bir filtreyi kapatin; REDUNDANS_UYGULA=False yapin; "
            f"ADIM 1'de LENGTH_MAX degerini yukseltin; "
            f"ya da N_EGITIM/N_TEST degerlerini dusurun.")

    if not TABAKALI_UZUNLUK:
        k = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
        return (k.iloc[:N_TEST].assign(split="test"),
                k.iloc[N_TEST:N_TEST + N_EGITIM].assign(split="train"))

    #  ORANTILI SEÇİM
    d = df.copy()
    d["_tab"] = pd.qcut(d["Length"], q=TABAKA_SAYISI, labels=False, duplicates="drop")
    tablar = sorted(d["_tab"].unique())
    paylar = np.array([(d["_tab"] == t).sum() for t in tablar], dtype=float)
    paylar /= paylar.sum()
    kota_te, kota_tr = _kota(paylar, N_TEST), _kota(paylar, N_EGITIM)

    te, tr, borc = [], [], 0
    for t, n_te, n_tr in zip(tablar, kota_te, kota_tr):
        alt = d[d["_tab"] == t].sample(frac=1.0, random_state=seed + int(t)) \
                               .reset_index(drop=True)
        al_te = min(n_te, len(alt))
        al_tr = min(n_tr, len(alt) - al_te)
        borc += (n_te - al_te) + (n_tr - al_tr)
        te.append(alt.iloc[:al_te])
        tr.append(alt.iloc[al_te:al_te + al_tr])
    te, tr = pd.concat(te), pd.concat(tr)

    if borc: 
        g(f"    ! [{sinif}] {borc} kayit tabaka disindan tamamlandi")
        secili = set(te["Uniprot_Accession"]) | set(tr["Uniprot_Accession"])
        kalan = d[~d["Uniprot_Accession"].isin(secili)] \
                  .sample(frac=1.0, random_state=seed).reset_index(drop=True)
        eksik_te, eksik_tr = N_TEST - len(te), N_EGITIM - len(tr)
        te = pd.concat([te, kalan.iloc[:eksik_te]])
        tr = pd.concat([tr, kalan.iloc[eksik_te:eksik_te + eksik_tr]])

    return (te.drop(columns="_tab").assign(split="test").reset_index(drop=True),
            tr.drop(columns="_tab").assign(split="train").reset_index(drop=True))


# ----------------------------- TANIMA ------------------------------------------
def uzunluk_auc(df: pd.DataFrame) -> float:


    x = df["Length"].astype(float)
    y = (df["label"] == "GPCR").values
    n1, n2 = int(y.sum()), int((~y).sum())
    if n1 == 0 or n2 == 0:
        return float("nan")
    r = x.rank()
    u1 = r[y].sum() - n1 * (n1 + 1) / 2
    auc = u1 / (n1 * n2)
    return float(max(auc, 1 - auc))


def fasta_yaz(df: pd.DataFrame, kolon: str, yol: str):
    with open(yol, "w") as f:
        for _, r in df.iterrows():
            f.write(f">{r['Uniprot_Accession']}|{r['label']}|{r['split']}\n{r[kolon]}\n")


# ----------------------------- ANA AKIŞ --------------------------------------
def main():
    g = Gunluk("ADIM 3 - ORNEKLEME")
    havuz = pd.read_excel(IN_XLSX, sheet_name="VERI")
    g(f"Girdi   : {IN_XLSX} | {len(havuz)} kayit")
    g(f"Ayarlar : N_EGITIM={N_EGITIM} N_TEST={N_TEST} SEED={SEED} "
      f"TABAKALI_UZUNLUK={TABAKALI_UZUNLUK} REDUNDANS_UYGULA={REDUNDANS_UYGULA}")
    g(f"Filtre  : CDS_DOGRU={SART_CDS_DOGRU} DOMAIN={SART_DOMAIN} "
      f"ALFABE={SART_ALFABE} FRAGMAN_HARIC={SART_FRAGMAN_HRC}")
    g("")

    parcalar = []
    for sinif, tohum in [("GPCR", SEED), ("KINASE", SEED + 1)]:
        g(f"[{sinif}] filtre hunisi:")
        alt = (filtre_uygula(havuz[havuz.label == sinif], g)
               .drop_duplicates("Uniprot_Accession").reset_index(drop=True))

        if REDUNDANS_OLC and len(alt) > 1:
            ey = redundans_olc(alt["AA_Sequence"].astype(str).tolist())
            g(f"    benzerlik (en yuksek k={K}-mer Jaccard): "
              f"ortanca={np.percentile(ey, 50):.3f} %90={np.percentile(ey, 90):.3f} "
              f"maks={ey.max():.3f} | {REDUNDANS_ESIK} ustu: {int((ey >= REDUNDANS_ESIK).sum())}")
        if REDUNDANS_UYGULA:
            alt = redundans_azalt(alt, REDUNDANS_ESIK, g)

        te, tr = bol(alt, sinif, tohum, g)
        g(f"    secilen: egitim {len(tr)} | test {len(te)} | kullanilmayan "
          f"{len(alt) - len(tr) - len(te)}")
        parcalar += [tr, te]
    g("")

    veri = pd.concat(parcalar, ignore_index=True)

    # ---  kontrol ---
    tr_set = set(veri.loc[veri.split == "train", "Uniprot_Accession"])
    te_set = set(veri.loc[veri.split == "test", "Uniprot_Accession"])
    assert not (tr_set & te_set), "Egitim/Test cakismasi!"
    assert veri["cds_len"].mod(3).eq(0).all(), "CDS uzunlugu 3'un kati degil!"
    assert veri["cds_dna"].str.upper().apply(lambda s: set(s) <= set("ACGT")).all(), \
        "CDS'te ACGT disi harf var!"

    veri = veri[[c for c in VERI_KOLON if c in veri.columns]]
    egitim = veri[veri.split == "train"].reset_index(drop=True)
    test = veri[veri.split == "test"].reset_index(drop=True)

    with pd.ExcelWriter(OUT_XLSX, engine="openpyxl") as w:
        egitim.to_excel(w, sheet_name="EGITIM", index=False)
        test.to_excel(w, sheet_name="TEST", index=False)

    if FASTA_YAZ:
        for ad, alt in [("egitim", egitim), ("test", test)]:
            fasta_yaz(alt, "AA_Sequence", f"{ad}_protein.fasta")
            fasta_yaz(alt, "cds_dna", f"{ad}_cds.fasta")

    # --- Özet  ---
    g("Bolum ozeti:")
    g(veri.groupby(["split", "label"])
      .agg(n=("Uniprot_Accession", "size"), aa_ort=("Length", "mean"),
           cds_ort=("cds_len", "mean"), cds_maks=("cds_len", "max"))
      .round(1).to_string())
    g("")
    if "alt_sinif" in veri.columns:
        g("Alt sinif dagilimi (ornekleme dengeyi bozdu mu):")
        g(pd.crosstab([veri["label"], veri["alt_sinif"]], veri["split"]).to_string())
        g("")
    g("Uzunluk (aa) ceyreklikleri:")
    g(veri.groupby(["split", "label"])["Length"]
      .describe(percentiles=[.25, .5, .75])[["min", "25%", "50%", "75%", "max"]]
      .round(1).to_string())
    g("")
    auc_tr, auc_te = uzunluk_auc(egitim), uzunluk_auc(test)
    g(f"TANI - uzunluk tek basina AUC: egitim={auc_tr:.3f} test={auc_te:.3f}")
    if auc_tr >= 0.70:
        g("  ! Uzunluk siniflar arasinda belirgin ayrisiyor. Sifir dolgulu girdide")
        g("    ag dizi yerine uzunlugu ogrenebilir; tezde tartisin ve sabit uzunluga")
        g("    kirpma ile duyarlilik analizi yapmayi degerlendirin.")
    else:
        g("  Uzunluk tek basina zayif ayirici; belirgin bir karistirici yok.")
    g("")
    g(f"Cikti  : {os.path.abspath(OUT_XLSX)} (EGITIM {len(egitim)} | TEST {len(test)})"
      + (" + FASTA" if FASTA_YAZ else ""))
    g.yaz()


if __name__ == "__main__":
    main()
```
